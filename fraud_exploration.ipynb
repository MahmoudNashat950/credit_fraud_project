```python
target = "Class"

# Correlation with target
corr_with_target = train_data.corr()[target]

# Sort by absolute correlation (excluding the target itself)
corr_sorted = corr_with_target.drop(target).reindex(
    corr_with_target.drop(target).abs().sort_values(ascending=False).index
)

# Pick top N features
top_n = 10
top_features = corr_sorted.index[:top_n]

# Add target at the end
selected_features = list(top_features) + [target]

# Build correlation matrix
corr_matrix = train_data[selected_features].corr()

# Plot heatmap
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, cmap="coolwarm", annot=True, fmt=".2f", square=True)
plt.title(f"Correlation Matrix of Top {top_n} Features (Target at Right)")
plt.show()
```

```python

from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, f1_score

# ==== 1. Define models ====
cat = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=6,
    scale_pos_weight=559,
    verbose=0,
    random_state=42
)

xgb = XGBClassifier(
    n_estimators=400,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=559,  # 170579 / 305
    eval_metric='logloss',
    random_state=42
)



lr = LogisticRegression(
    class_weight={0:1, 1:15},
    solver='liblinear',
    random_state=42
)
# ==== 2. Combine in a soft-voting ensemble ====
voting_clf = VotingClassifier(
    estimators=[('lr', lr), ('xgb', xgb), ('cat', cat)],
    voting='soft'
)


# ==== 3. Train the model ====
voting_clf.fit(X_train_scaled, y_train)

# ==== 4. Evaluate ====

print("\n--- Training Set ---")
y_train_pred = voting_clf.predict(X_train_scaled)
print(classification_report(y_train, y_train_pred))
print("ROC-AUC:", roc_auc_score(y_train, y_train_pred))

print("\n--- Validation Set ---")
y_val_pred = voting_clf.predict(X_val_scaled)
print(classification_report(y_val, y_val_pred))
print("ROC-AUC:", roc_auc_score(y_val, y_val_pred))

print("\n--- Test Set ---")
y_test_pred = voting_clf.predict(X_test_scaled)
print(classification_report(y_test, y_test_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_test_pred))
```

```python
# Get probabilities for each class
y_proba_test = voting_clf.predict_proba(X_test_scaled)[:, 1]  # probability of class 1
y_proba_val = voting_clf.predict_proba(X_val_scaled)[:, 1]  # probability of class 1
y_proba_train = voting_clf.predict_proba(X_train_scaled)[:, 1]  # probability of class 1

best_threshold = 0.7445532313170999
y_pred_test = (y_proba_test >= best_threshold).astype(int)
y_pred_val = (y_proba_val >= best_threshold).astype(int)
y_pred_train = (y_proba_train >= best_threshold).astype(int)
```

let's find the best threshold

```python

from sklearn.metrics import precision_recall_curve, auc

# Get probabilities for class 1 (fraud)
y_scores = voting_clf.predict_proba(X_test_scaled)[:, 1]

# Compute precision-recall pairs
precision, recall, thresholds = precision_recall_curve(y_test, y_scores)

# Compute AUC for PR curve
pr_auc = auc(recall, precision)

# Plot curve
plt.figure(figsize=(8,6))
plt.plot(recall, precision, marker='.', label=f'PR curve (AUC = {pr_auc:.2f})')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.metrics import f1_score

f1_scores = []
for t in thresholds:
    y_pred_thresh = (y_scores >= t).astype(int)
    f1_scores.append(f1_score(y_test, y_pred_thresh))

best_idx = f1_scores.index(max(f1_scores))
best_threshold = thresholds[best_idx]

print("Best threshold:", best_threshold)
print("Best F1-score:", max(f1_scores))
```
